## Attach the reviewed dataset
Attach the private ground-level pilot containing manifest.csv, split.csv, images/ and ATTRIBUTION.txt. The supplied split must be preserved, including GIGO original test placement. This small pilot cannot establish Philippine production accuracy.


```python
import subprocess, shutil
from barangai_ml.cli import read_split
candidates = [p.parent for p in Path('/kaggle/input').rglob('manifest.csv') if (p.parent/'split.csv').is_file()]
assert len(candidates) == 1, f'Expected one reviewed dataset, found {candidates}'
DATASET = candidates[0]
IMAGES = DATASET
SPLIT = WORK/'outputs'/'split.csv'
OUTPUT = WORK/'outputs'/'comparison'
SPLIT.parent.mkdir(parents=True, exist_ok=True)
records, splits = read_split(DATASET/'split.csv', IMAGES)
shutil.copyfile(DATASET/'split.csv', SPLIT)
print('Reviewed dataset:', DATASET)
print('Verified images:', len(records), {s:splits.count(s) for s in ('train','val','test')})
assert len(records) == 75, 'Unexpected dataset revision: review before training.'
```

## Train both models
This trains both ImageNet-initialized architectures on the same fixed split. Choose model and uncertainty threshold using validation results; report held-out test results without retuning. Ten epochs are an initial experiment, not a promised accuracy target.

```python
subprocess.run([sys.executable,'-m','barangai_ml.cli','train','--split',str(SPLIT),
                '--image-root',str(IMAGES),'--output',str(OUTPUT),'--epochs','10',
                '--batch-size','16','--seed','42'],cwd=WORK,check=True)
print((OUTPUT/'results.json').read_text())
```
